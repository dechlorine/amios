```python
from random import random
from math import log, floor, ceil
from statistics import mean, stdev
import matplotlib.pyplot as plt
```

```python
def random_uniform():
    x = random()
    return x * 200

def random_exp():
    x = random()
    return log(1 - x) * (-110)

def random_triangle():
    x = random()
    return 20 - 20 * (1 - x)**0.5
```

```python
def generate(N, func):
    nums = []
    for x in range(N):
        num = func()
        nums.append(num)
    return nums
```

```python
def bar_chart(nums, step, n=20):
    heights = [0] * n
    for i in range(len(nums)):
        heights[floor(nums[i] / step)] += 1

    centers = [i * step + step // 2 for i in range(n)]
    plt.bar(centers, heights, width=step, edgecolor="black", align="center")

    boundaries = [i * step for i in range(n + 1)]
    plt.xticks(boundaries, rotation=45, ha="right")

    plt.tight_layout()
    plt.show()
```

```python
lens = [100, 1000, 10000]
uni_nums = []
exp_nums = []
tri_nums = []
for i in lens:
  uni_nums.append(generate(i, random_uniform))
  exp_nums.append(generate(i, random_exp))
  tri_nums.append(generate(i, random_triangle))
```

```python
for i in range(3):
  bar_chart(uni_nums[i], step=10, n=20)
```

```python
for i in range(3):
  bar_chart(exp_nums[i], step=ceil(max(exp_nums[i])/200)*10, n=20)
```

```python
for i in range(3):
  bar_chart(tri_nums[i], step=1, n=20)
```

```python
print("MEAN theor:", 100)
for i in range(3):
  print("MEAN pract:", mean(uni_nums[i]))

print("\n")

print("STDEV theor:", (200*3**0.5)/6)
for i in range(3):
  print("STDEV pract:", stdev(uni_nums[i]))
```

```python
print("MEAN theor:", 110)
for i in range(3):
  print("MEAN pract:", mean(exp_nums[i]))

print("\n")

print("STDEV theor:", 110)
for i in range(3):
  print("STDEV pract:", stdev(exp_nums[i]))
```

```python
print("MEAN theor:", 20/3)
for i in range(3):
  print("MEAN pract:", mean(tri_nums[i]))

print("\n")

print("STDEV theor:", (20*20/18)**0.5)
for i in range(3):
  print("STDEV pract:", stdev(tri_nums[i]))
```
